```python
finalbook_list = [] #empty list for storing books details as dictionary

#function for login to LMS system---Step 5 according to project
def credential():
    #Login system
    user="admin"
    keyword="admin@123"
    i=0
    while i<=2:
        user_id=input("Enter login User ID")
        password=input("Enter user password")
        if (user_id==user and password==keyword):
            print("\nWelcome to library management System:-")
            main_menu()
            break
            # return True
            # break
        elif i==2:
            print("\nCheck credentials and come back again")
            exit()
            #return False
            break
            
        else:
            print("\nwrong credentials")
            i=i+1
        
# Function to add book details---Step 2 according to project
def book_add():
    key = ['Name', 'Author', 'Available', 'Issued_By']
    a = input('Enter book name: ')  #Name of the book
    for book in finalbook_list:
        if a == book['Name']:  #Check if book name already exists
            print("Book already exists")
            return None 
    
    b = input('Enter author: ')  #Author of the book
    c_input = input('Enter availability (True/False): ')  #Availability
    c = c_input.lower() == 'true' 
    
    # Creating a dictionary of book details
    book_details = dict(zip(key, [a, b, c, None]))
    print(book_details)  #Printing book details.
    print(f'Details of book added are: Book Name: {book_details["Name"]}, Author: {book_details["Author"]}, Availability: {book_details["Available"]}')
    return book_details

#Function for issuing books---Step 3 according to project
def issue_book():
    book_name = input('Enter the name of the book you want to issue: ')
    
    #verifying if the book available
    for book in finalbook_list:
        if book['Name'].lower() == book_name.lower():
            if book['Available']:
                name_of_person = input('Enter your name: ')  
                book['Available'] = False  
                book['Issued_By'] = name_of_person
                print(f"The book '{book_name}' has been issued to {name_of_person}.")
                return
            else:
                print(f"Sorry, the book '{book_name}' is already issued.")
                return
    
    #If the book does not exist
    print(f"Sorry, the book '{book_name}' is not available in the list.")
    
#functtion to return a book---Step 4 according to project
def return_book():
    book_name=input("Enter the name of book to be returned")
    for book in finalbook_list:
        if book['Name'].lower() == book_name.lower(): #if book exists in library database
            if book['Available']==False:
                book['Available'] = True
                print("Book returned back successfully")
        else:
            print("Book not of library")

#function for viewing issued books ---Step 6 according to project           
def issued_books():
     if finalbook_list:
                print("\nHere is the Issued books list:-")
                for index, book in enumerate(finalbook_list, start=1):
                    if book['Available'] == False:
                        print(f"{index}. {book['Name']} issued to {book['Issued_By']}")
                    else:
                        print("\nNo books issued!")

#Main Menu function----Step 7 according tot project
def main_menu():
    
    while True:
        print("\n--- Main Menu ---")
        print("1. Add a new book")
        print("2. Show all books")
        print("3. Issue a book")
        print("4. Return book")
        print("5. view issued books")
        print("6. Exit")
        
        choice = input("Please choose an option (1/2/3/4/5/6): ")
        
        if choice == '1':
            #Add a new book
            book = book_add()
            if book:  # Only add the book if it was successfully added
                finalbook_list.append(book)  #updating final bok list
                print("Book added successfully!\n")
        
        elif choice == '2':
            #show all books
            if finalbook_list:
                print("\nHere is the updated list of all books:")
                for index, book in enumerate(finalbook_list, start=1):
                    status = "Available" if book['Available'] else f"Issued to {book['Issued_By']}"
                    print(f"{index}. {book['Name']} by {book['Author']}, Status: {status}")
            else:
                print("\nNo books added")
        
        elif choice == '3':
            #Issue a book
            issue_book()
            
        elif choice == '4':
            #Return a book
            return_book()
            
        elif choice == '5':
            #view issued books
            issued_books()
                    
        elif choice == '6':
            #Exit the program
            print("Exiting the program. Goodbye!")
            break
        
        else:
            print("Invalid choice. Please try again.")

# start with credentials to start the program
credential()
```
